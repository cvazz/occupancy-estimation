```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, filters, img_as_float
from scipy.ndimage import gaussian_filter

# Load the camera image
image = img_as_float(data.camera())

# Apply two Gaussian blurs with different sigmas
sigma1 = 1
sigma2 = 3
blur1 = gaussian_filter(image, sigma=sigma1)
blur2 = gaussian_filter(image, sigma=sigma2)

# Difference of Gaussians
dog = blur1 - blur2

# Plotting
fig, axes = plt.subplots(1, 4, figsize=(16, 5))
axes[0].imshow(image, cmap='gray')
axes[0].set_title("Original Image")
axes[0].axis('off')

axes[1].imshow(blur1, cmap='gray')
axes[1].set_title(f"Gaussian Blur σ={sigma1}")
axes[1].axis('off')

axes[2].imshow(blur2, cmap='gray')
axes[2].set_title(f"Gaussian Blur σ={sigma2}")
axes[2].axis('off')

axes[3].imshow(dog, cmap='gray')
axes[3].set_title("DoG Result")
axes[3].axis('off')

plt.tight_layout()
plt.show()
```
